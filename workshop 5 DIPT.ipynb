```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('car.png')
```

```python
def display(img):
    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111)
    ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.show()
```

```python
display(img)
```

```python
plate_cascade = cv2.CascadeClassifier('haarcascade_licence_plate_rus_16stages.xml')
```

```python
def detect_plate(img):
    img_copy = img.copy()
    
    plate_rects = plate_cascade.detectMultiScale(img_copy, scaleFactor=1.1, minNeighbors=5)
    
    for (x, y, w, h) in plate_rects:
        cv2.rectangle(img_copy, (x, y), (x + w, y + h), (0, 0, 255), 3)
    
    return img_copy
```

```python
result = detect_plate(img)
```

```python
display(result)
```

```python
def detect_and_blur_plate(img):
    img_copy = img.copy()
    
    plate_rects = plate_cascade.detectMultiScale(img_copy, scaleFactor=1.1, minNeighbors=5)
    
    for (x, y, w, h) in plate_rects:
        roi = img_copy[y:y+h, x:x+w]
        roi = cv2.medianBlur(roi, 15)
        img_copy[y:y+h, x:x+w] = roi
    
    return img_copy
```

```python
result = detect_and_blur_plate(img)
```

```python
display(result)
```

```python
def detect_plate(img):
    img_copy = img.copy()

    gray = cv2.cvtColor(img_copy, cv2.COLOR_BGR2GRAY)

    plates = plate_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=4
    )

    for (x, y, w, h) in plates:
        cv2.rectangle(
            img_copy,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            3
        )

    return img_copy
```

```python
detected = detect_plate(img)
display(detected)
```

```python
def detect_and_blur_plate(img):
    img_copy = img.copy()

    gray = cv2.cvtColor(img_copy, cv2.COLOR_BGR2GRAY)

    plates = plate_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=4
    )

    for (x, y, w, h) in plates:
        roi = img_copy[y:y+h, x:x+w]

        blurred_roi = cv2.medianBlur(roi, 15)

        img_copy[y:y+h, x:x+w] = blurred_roi

    return img_copy
```

```python
result = detect_and_blur_plate(img)
display(result)
```
